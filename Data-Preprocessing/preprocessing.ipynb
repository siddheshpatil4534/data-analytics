# Data Preprocessing

## Objective

The objective of this notebook is to perform basic data preprocessing on the Titanic dataset using Python. The preprocessing includes handling missing values, feature scaling, encoding categorical data, and detecting outliers.

```python
# Import required libraries

import pandas as pd
```

```python
# Load dataset

df = pd.read_csv("data/tested.csv")

print(df.head())
```

## Dataset Information

```python
print(df.info())

print(df.columns)
```

## Handling Missing Values

```python
# Number of missing values in Age

print(df["Age"].isnull().sum())
```

```python
# Calculate mean Age

mean_age = df["Age"].mean()

print(mean_age)
```

```python
# Fill missing Age values with mean

df["Age"] = df["Age"].fillna(mean_age)

print(df["Age"].isnull().sum())
```

```python
print(df.head())
```

## Display Complete Dataset

```python
pd.set_option("display.max_columns", None)

print(df.head())
```

## Target Variable Distribution

```python
print(df["Survived"].value_counts())
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df["Age"] = scaler.fit_transform(df[["Age"]])

print(df[["Age"]].head())
```

## Encoding Categorical Data

```python
df["sex"] = df["Sex"].map({"male": 0, "female": 1})

print(df[["Sex", "sex"]].head())
```

## Outlier Detection using IQR

```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1
```

```python
outliers = df[
    (df["Age"] < Q1 - 1.5 * IQR) |
    (df["Age"] > Q3 + 1.5 * IQR)
]

print(outliers[["Name", "Age"]])
```

```python
print("Number of outliers:", len(outliers))
```

## Observations

- Missing values in the Age column were replaced with the mean.
- Age values were standardized using StandardScaler.
- The Sex column was encoded into numerical values.
- Outliers in the Age column were detected using the IQR method.

## Conclusion

The dataset was preprocessed by handling missing values, scaling numerical data, encoding categorical values, and detecting outliers. These preprocessing steps make the dataset more suitable for machine learning models.
