# Exploratory Data Analysis (EDA) on Iris Dataset

## Objective

# The objective of this notebook is to perform Exploratory Data Analysis (EDA) on the Iris dataset using Python. The analysis includes dataset exploration, descriptive statistics, histogram visualization, and correlation analysis to understand the characteristics of the data.

```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inside the notebook
%matplotlib inline

# Improve plot appearance
sns.set_theme(style="whitegrid")
```

```python
 #data set loading
df = pd.read_csv("data/Iris.csv")
```

```python
# Display the first five rows
df.head()
```

# Dataset information

```python
# Shape of the dataset
print("Shape of the dataset:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# Dataset information
df.info()
```

## Descriptive Statistics

```python
# Descriptive statistics of SepalLengthCm

df["SepalLengthCm"].describe()
```

## Mean, Median and Mode

```python
# Calculate mean, median and mode

mean = df["SepalLengthCm"].mean()
median = df["SepalLengthCm"].median()
mode = df["SepalLengthCm"].mode()[0]

print("Mean   :", mean)
print("Median :", median)
print("Mode   :", mode)
```

## Histogram

```python
plt.hist(df["SepalLengthCm"], bins=10, edgecolor="black")

plt.axvline(mean, color="red", label="Mean")
plt.axvline(median, color="green", label="Median")
plt.axvline(mode, color="orange", label="Mode")

plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

## Correlation Matrix

```python
correlation = df.corr(numeric_only=True)

correlation
```

```python
sns.heatmap(correlation, annot=True)

plt.title("Correlation Matrix")
plt.show()
```

## Observations

- The Iris dataset contains 150 records.
- The dataset has no missing values.
- The mean, median and mode of Sepal Length were calculated.
- The histogram shows the distribution of Sepal Length.
- Petal Length and Petal Width have a strong positive correlation.
- Sepal Width has a weak negative correlation with Petal Length and Petal Width.

## Conclusion

Exploratory Data Analysis was successfully performed on the Iris dataset. The dataset was analyzed using descriptive statistics, histogram visualization, and correlation analysis. The results provide a better understanding of the data and its relationships.
