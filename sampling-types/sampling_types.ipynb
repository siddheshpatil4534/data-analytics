# Sampling Types

Sampling is the process of selecting a smaller group of observations from a larger dataset.

In this notebook, we will learn and implement:
- Simple Random Sampling
- Stratified Sampling
- Systematic Sampling

```python
import pandas as pd
import numpy as np
```

## Loading the Dataset

We will use the student performance dataset and first look at its structure before applying different sampling methods.

```python
df = pd.read_csv("students_performance.csv")

df.head()
```

## Understanding the Dataset

Before sampling, let's check how many records and columns are available in the dataset.

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.columns
```

## Simple Random Sampling

Randomly select 100 values from the `math score` column.

```python
random_sample = df["math score"].sample(n=100, random_state=42)

random_sample.head()
```

### Sample Size

```python
print("Population size:", len(df["math score"]))
print("Sample size:", len(random_sample))
```

## Stratified Sampling

Divide the students into groups based on their `gender` and select samples from each group.

```python
stratified_sample = (
    df.groupby("gender", group_keys=False)
      .apply(lambda x: x.sample(n=50, random_state=42))
)

stratified_sample["math score"].head()
```

## Systematic Sampling

Select every fixed interval from the `math score` column.

```python
interval = 10

systematic_sample = df["math score"].iloc[::interval]

systematic_sample.head(10)
```

## Comparing the Samples

Compare the mean and spread of the three sampling methods.

```python
comparison = pd.DataFrame({
    "Simple Random": random_sample.describe(),
    "Stratified": stratified_sample["math score"].describe(),
    "Systematic": systematic_sample.describe()
})

comparison
```
