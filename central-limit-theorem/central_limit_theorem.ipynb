# Central Limit Theorem

The Central Limit Theorem shows how the distribution of sample means approaches a normal distribution as the number of samples increases.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Loading the Dataset

We will use the `math score` column.

```python
df = pd.read_csv("students_performance.csv")

scores = df["math score"]
scores.head()
```

## Before Applying CLT

Let's look at the original distribution of `math score`.

```python
plt.hist(scores, bins=20)
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.title("Original Math Score Distribution")
plt.show()
```

## Applying the Central Limit Theorem

Take many samples of 30 scores and calculate the mean of each sample.

```python
sample_means = []

for _ in range(1000):
    sample = np.random.choice(scores, size=30)
    sample_means.append(np.mean(sample))

sample_means = np.array(sample_means)
```

## After Applying CLT

The distribution of the sample means should be closer to a normal distribution.

```python
plt.hist(sample_means, bins=20)
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("Distribution of Sample Means")
plt.show()
```

## Comparing Before and After

The original scores show the population distribution, while the sample means become more normally distributed after applying CLT.
